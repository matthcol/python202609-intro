### Temporal Types

```python
from datetime import date, datetime, time
from zoneinfo import ZoneInfo # dictionnaire IANA
import locale
```

```python
d1 = date.today()
d2 = date(2024, 2, 29)
d3 = date.strptime("04/09/2026", "%d/%m/%Y")
print(d1)
print(d2)
print(d3)
```

```python
tz1 = ZoneInfo('Europe/Paris')
tz2 = ZoneInfo('America/New_York')
tz3 = ZoneInfo('Asia/Tokyo')
dt0 = datetime.now()
dt1 = datetime.now(tz1)
dt2 = datetime.now(tz=tz2)
dt3 = datetime.now(tz=tz3)
for dt in dt0, dt1, dt2, dt3:
    print(dt)
    print(dt.date())
    print(dt.time())
    print(f"{dt:%d/%m/%Y à %H:%M}")
    print(f"{dt:%A %d %B %Y à %H:%M}")
    print()
```

```python
locale.setlocale(locale.LC_ALL, 'fr-FR.UTF-8')

for dt in dt0, dt1, dt2, dt3:
    print(f"{dt:%A %d %B %Y à %H:%M %z}")

print()

for lieu, dt in ('Ici', dt0), ('Paris', dt1), ('New York', dt2), ('Tokyo', dt3):
    print(f"À {lieu}, {dt:%A %d %B %Y à %H:%M %z}")
```

```python
city_info_list = [
    ('Toulouse', '31000', 500_000),
    ('Pau', '64000', 77_000),
    ('Argenteuil', '95100', 100_000),
    ('Villejuif', '94800', 60_000),
]

city_info_list.append(('Nice', '06000', 300_000))
# city_info_list.append(('Paris', 2_100_000, '75000'))
# city_info_list.append(('Paris', '75000', 2_100_000, 'Capitale'))

for city_info in city_info_list:
    print(city_info)
```

```python
print("Villes : ")
for name, cp, pop in city_info_list:
    print(f" * Ville = {name}, code postal principal = {cp}, population = {pop}, département = {cp[:2]}")
```

```python
total_pop = 0
for *_, pop in city_info_list:
    total_pop += pop
print(f"Population totale des {len(city_info_list)} villes : {total_pop}")

# variante avec une expression for (generator expression)
total_pop = sum(pop for *_, pop in city_info_list)
print(f"Population totale des {len(city_info_list)} villes : {total_pop}")
```

### Fonctions

```python
def search_city(dfCities: pd.DataFrame, nom: str, departement: str | None = None) -> pd.DataFrame:
    mask = (
        dfCities['nom_standard'].str.contains(nom, case=False, regex=True, na=False)
        | dfCities['nom_sans_accent'].str.contains(nom, case=False, regex=True, na=False)
    )
    if departement is not None:
        mask &= dfCities['dep_code'] == departement
    return dfCities.loc[mask]
```

```python
search_city(dfCitiesFr, 'paris')
```

```python
search_city(dfCitiesFr, '^paris')
```

```python
search_city(dfCitiesFr, 'hay-les-roses')
```

```python
search_city(dfCitiesFr, 'paris', '81')
```

```python
search_city(dfCitiesFr, 'paris', departement='81')
```

## Type custom

```python
from dataclasses import dataclass


@dataclass
class CityFr:
    name: str
    zipcode: str
    population: int


citiesFrList = [
    CityFr('Paris', '75000', 2_100_000),
    CityFr('Toulouse', '31000', 500_000),
    CityFr('Pau', '64000', 77_000),
]

for city in citiesFrList:
    print(city)
```

```python
from pydantic import BaseModel, Field  # Note: validation des données, serialisation JSON


class CityFr(BaseModel):
    name: str = Field(min_length=1)
    zipcode: str = Field(min_length=5, max_length=5)
    population: int = Field(ge=0)


citiesFrList = [
    CityFr(name='Paris', zipcode='75000', population=2_100_000),
    CityFr(name='Toulouse', zipcode='31000', population=500_000),
    CityFr(name='Pau', zipcode='64000', population=77_000),

    # CityFr(name='', zipcode='64000', population=77_000),
    # ValidationError: 1 validation error for CityFr
    # name
    # String should have at least 1 character [type=string_too_short, input_value='', input_type=str]

    # CityFr(name='Pau', zipcode='6400', population=77_000),
    # ValidationError: 1 validation error for CityFr
    # zipcode
    # String should have at least 5 characters [type=string_too_short, input_value='6400', input_type=str]
    #     For further information visit https://errors.pydantic.dev/2.13/v/string_too_short

    # CityFr(name='Pau', zipcode='64000', population=-1),
    # ValidationError: 1 validation error for CityFr
    # population
    # Input should be greater than or equal to 0 [type=greater_than_equal, input_value=-1, input_type=int]
    #     For further information visit https://errors.pydantic.dev/2.13/v/greater_than_equal
]

for city in citiesFrList:
    print("* City :     ", city)
    print("   => JSON : ", city.model_dump_json())
```
